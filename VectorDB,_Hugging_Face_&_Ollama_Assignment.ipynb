Question 1: What is a Vector Database (VectorDB) and how is it different from
traditional databases?

    Answer:A Vector Database (VectorDB) is a specialized database designed to store, manage, and search vector embeddings. Vector embeddings are numerical representations of data such as text, images, audio, or documents. Similar or related data have vectors that are close to each other in the vector space.

    Vector databases are widely used in AI and Generative AI applications, especially for semantic search, RAG (Retrieval-Augmented Generation), recommendation systems, and question-answering systems.


  | Feature                | Traditional Database                     | Vector Database                                    |
  | ---------------------- | ---------------------------------------- | -------------------------------------------------- |
  | **Data stored**        | Structured data such as rows and columns | Numerical vectors/embeddings                       |
  | **Search method**      | Exact matching, filters, SQL queries     | Similarity/semantic search                         |
  | **Search basis**       | Keywords, values, conditions             | Meaning and similarity                             |
  | **Example query**      | `WHERE age = 25`                         | Find documents similar to a given question         |
  | **Common use**         | Banking, employee records, inventory     | RAG, AI search, recommendations                    |
  | **Similarity metrics** | Usually not the main focus               | Cosine similarity, Euclidean distance, dot product |
  | **Examples**           | MySQL, PostgreSQL, Oracle                | ChromaDB, FAISS, Pinecone, Milvus, Weaviate        |


Question 2: Explain the various types of VectorDBs available and describe their
suitability for different use cases.
    
    Answer:A Vector Database (VectorDB) can be classified based on how it stores, searches, and manages vector embeddings. Different VectorDBs are suitable for different applications depending on factors such as scale, speed, cloud requirements, filtering, and ease of deployment.

Types of VectorDBs and Their Use Cases:

| VectorDB          | Type / Feature                                      | Suitable Use Cases                                                  |
| ----------------- | --------------------------------------------------- | ------------------------------------------------------------------- |
| **FAISS**         | Open-source vector similarity search library        | Local projects, experimentation, research, small-to-medium datasets |
| **ChromaDB**      | Developer-friendly vector database                  | RAG applications, chatbots, document search, learning projects      |
| **Pinecone**      | Fully managed cloud VectorDB                        | Production AI applications, scalable RAG, recommendation systems    |
| **Weaviate**      | Open-source and cloud vector database               | Semantic search, RAG, multimodal applications                       |
| **Milvus**        | Open-source, highly scalable VectorDB               | Large-scale enterprise applications, billions of vectors            |
| **Qdrant**        | Vector search engine with strong metadata filtering | RAG, recommendation systems, semantic search                        |
| **Elasticsearch** | Search engine with vector-search capabilities       | Applications requiring both keyword and semantic search             |
| **pgvector**      | Vector extension for PostgreSQL                     | Applications already using PostgreSQL and requiring vector search   |

Question 3: Why is Chroma DB considered important in the context of AI/ML projects?
Describe its key features.
    
    Answer:ChromaDB is an open-source vector database designed to store, manage, and search vector embeddings. It is especially useful in AI/ML and Generative AI projects because it makes it easy to retrieve information based on semantic similarity rather than only exact keyword matching.

    Key Features of ChromaDB

    1. Vector Storage
    ChromaDB stores vector embeddings generated by embedding models.

    2. Similarity Search
    It can find vectors that are most similar to a given query vector, allowing AI applications to retrieve relevant information.

    3. Metadata Filtering
    Documents can be stored with metadata such as source, category, author, or date. This metadata can be used to filter search results.

    4. Easy Integration
    ChromaDB integrates easily with popular AI/ML frameworks such as LangChain and embedding models.

    5. Persistent Storage
    ChromaDB can persist collections so that the stored vectors do not have to be recreated every time the application starts.

    6. Supports RAG Applications
    In a RAG system, ChromaDB can store document embeddings and retrieve relevant document chunks before sending them to an LLM.

    7. Developer-Friendly
    It has a relatively simple API, making it convenient for beginners and developers building prototypes and AI applications.

```python
# Question 7: Demonstrate how to download and fine-tune a Hugging Face model for a
# text generation task.
# Install first:
# pip install transformers datasets accelerate torch

from datasets import Dataset
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling,
    pipeline
)

# --------------------------------------------------
# 1. Create sample dataset
# --------------------------------------------------

texts = [
    "Artificial Intelligence enables computers to perform tasks that normally require human intelligence.",
    "Machine learning is a branch of Artificial Intelligence that learns patterns from data.",
    "Deep learning uses neural networks with multiple layers to learn complex patterns.",
    "Natural Language Processing allows computers to understand and generate human language.",
    "Generative AI can create text, images, audio, and other types of content.",
    "Large Language Models are trained on large amounts of text data.",
    "Python is widely used for Artificial Intelligence and machine learning projects.",
    "AI chatbots use language models to understand questions and generate responses."
]

dataset = Dataset.from_dict({
    "text": texts
})

print("Dataset:")
print(dataset)

# --------------------------------------------------
# 2. Download pre-trained GPT-2
# --------------------------------------------------

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(model_name)

# GPT-2 does not have a padding token
tokenizer.pad_token = tokenizer.eos_token
model.config.pad_token_id = tokenizer.pad_token_id

# --------------------------------------------------
# 3. Tokenize dataset
# --------------------------------------------------

def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=128
    )

tokenized_dataset = dataset.map(
    tokenize_function,
    batched=True
)

# --------------------------------------------------
# 4. Data collator
# --------------------------------------------------

data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

# --------------------------------------------------
# 5. Training configuration
# --------------------------------------------------

training_args = TrainingArguments(
    output_dir="./gpt2-finetuned",
    num_train_epochs=3,
    per_device_train_batch_size=2,
    learning_rate=5e-5,
    logging_steps=1,
    save_strategy="epoch",
    report_to="none"
)

# --------------------------------------------------
# 6. Create Trainer
# --------------------------------------------------

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
    data_collator=data_collator
)

# --------------------------------------------------
# 7. Fine-tune
# --------------------------------------------------

print("\nStarting fine-tuning...")

trainer.train()

# --------------------------------------------------
# 8. Save fine-tuned model
# --------------------------------------------------

save_path = "./gpt2-ai-model"

trainer.save_model(save_path)
tokenizer.save_pretrained(save_path)

print("\nModel saved to:", save_path)

# --------------------------------------------------
# 9. Load fine-tuned model
# --------------------------------------------------

generator = pipeline(
    "text-generation",
    model=save_path,
    tokenizer=save_path
)

# --------------------------------------------------
# 10. Generate text
# --------------------------------------------------

prompt = "Artificial Intelligence"

result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1
)

print("\nGenerated Text:")
print(result[0]["generated_text"])
```

```python
# Question 9: Implement a basic RAG (Retrieval-Augmented Generation) system using
# Ollama with Llama3.
import ollama
import chromadb

# -----------------------------------------
# 1. Create sample documents
# -----------------------------------------

documents = [
    "Artificial Intelligence is a field of computer science that enables machines to perform tasks that normally require human intelligence.",

    "Machine Learning is a subset of Artificial Intelligence that allows computers to learn patterns from data without being explicitly programmed for every task.",

    "Deep Learning is a branch of Machine Learning that uses artificial neural networks with multiple layers to learn complex patterns.",

    "Generative AI is a type of artificial intelligence that can generate new content such as text, images, audio, and code.",

    "Retrieval-Augmented Generation combines information retrieval with a language model to generate answers using external knowledge."
]

# -----------------------------------------
# 2. Create ChromaDB client
# -----------------------------------------

client = chromadb.Client()

# -----------------------------------------
# 3. Create collection
# -----------------------------------------

collection = client.create_collection(
    name="ai_knowledge"
)

# -----------------------------------------
# 4. Create embeddings and store documents
# -----------------------------------------

for i, document in enumerate(documents):

    response = ollama.embeddings(
        model="nomic-embed-text",
        prompt=document
    )

    embedding = response["embedding"]

    collection.add(
        ids=[str(i)],
        documents=[document],
        embeddings=[embedding]
    )

# -----------------------------------------
# 5. User question
# -----------------------------------------

question = "What is Retrieval-Augmented Generation?"

# -----------------------------------------
# 6. Create embedding for question
# -----------------------------------------

question_embedding = ollama.embeddings(
    model="nomic-embed-text",
    prompt=question
)["embedding"]

# -----------------------------------------
# 7. Search ChromaDB
# -----------------------------------------

results = collection.query(
    query_embeddings=[question_embedding],
    n_results=2
)

retrieved_documents = results["documents"][0]

print("\nRetrieved Documents:")

for doc in retrieved_documents:
    print("-", doc)

# -----------------------------------------
# 8. Create context
# -----------------------------------------

context = "\n\n".join(retrieved_documents)

# -----------------------------------------
# 9. Create RAG prompt
# -----------------------------------------

prompt = f"""
You are a helpful AI assistant.

Answer the question using only the provided context.

Context:
{context}

Question:
{question}

Answer:
"""

# -----------------------------------------
# 10. Generate answer using Llama 3
# -----------------------------------------

response = ollama.chat(
    model="llama3",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ]
)

# -----------------------------------------
# 11. Print final answer
# -----------------------------------------

print("\nFinal Answer:")
print(response["message"]["content"])
```

The system will use Hugging Face models for understanding and embeddings, a VectorDB for retrieving relevant research articles, and Ollama for generating the final response locally.

    Proposed Architecture
    Medical Research Articles
              ↓
      Text Extraction
              ↓
          Chunking
              ↓
    Hugging Face Embedding Model

        Vector Embeddings
              ↓
          VectorDB
              ↓
          User Question
              ↓

              ↓

    1. Collect Medical Research Articles

    The startup can collect relevant medical research papers from legitimate sources and store them in formats such as PDF, HTML, or text.

    Each document should contain useful metadata such as:

    Article title
    Authors
    Publication date
    Journal
    Research topic
    Source URL
    DOI, where available

    This metadata helps with filtering and displaying citations.

    2. Process and Split the Documents

    Large research papers should be divided into smaller chunks.

    For example:

    Research Paper
          ↓
    Introduction
          ↓
    Methods
          ↓
    Results
          ↓
    Discussion
          ↓
    Smaller text chunks

    Chunking makes it easier for the retrieval system to find the specific section relevant to a user's question.

    3. Use Hugging Face Models for Understanding

    A Hugging Face embedding model can convert each text chunk into a numerical vector.

    For example:

    from sentence_transformers import SentenceTransformer

    embedding = model.encode(
        "Research shows that regular exercise can improve cardiovascular health."

    print(embedding.shape)

    The same embedding model should be used to convert the user's question into a vector.

    The system can then compare the question vector with the document vectors.

    4. Store Embeddings in a VectorDB

    The generated embeddings can be stored in a VectorDB such as ChromaDB, Qdrant, Pinecone, or Milvus.

    For example, with ChromaDB:

    import chromadb

    The VectorDB allows the system to perform semantic search rather than relying only on exact keywords.

    5. Retrieve Relevant Research

    Suppose the user asks:

    "What does research say about exercise and cardiovascular health?"

    The question is converted into an embedding.

    The VectorDB searches for the most similar research chunks:

    results = collection.query(
        query_embeddings=[question_embedding],
